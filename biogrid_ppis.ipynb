## Cross-reference BioGRID interactors against the S. pombe gene list

For every gene that appears as an interactor in BioGRID (either Interactor A or B), check whether it is present in the full S. pombe gene list and, if so, record its `gene_category` (Studied, Priority Unstudied, or Non-Priority Unstudied).

```python
# Load full BioGRID dataset
biogrid_path = BASE_DIR / "ppis" / "biogrid" / "BIOGRID_pombe_physical_interactions.csv"
biogrid_df = pd.read_csv(biogrid_path, sep="\t")
print(f"Total BioGRID interactions: {len(biogrid_df):,}")

# Load pombe gene list (read-only)
pombe_genes = pd.read_csv(pombe_gene_list_path, sep="\t")
pombe_gene_set = set(pombe_genes["gene_systematic_id"])
print(f"Total S. pombe genes in gene list: {len(pombe_gene_set):,}")

# Collect every unique systematic name that appears in either interactor column
all_interactors = pd.concat([
    biogrid_df["Systematic.Name.Interactor.A"].rename("systematic_id"),
    biogrid_df["Systematic.Name.Interactor.B"].rename("systematic_id"),
]).dropna().unique()
all_interactors = pd.Series(all_interactors, name="systematic_id")
print(f"\nUnique interactor IDs in BioGRID: {len(all_interactors):,}")

# Match interactors against the pombe gene list
matched = all_interactors[all_interactors.isin(pombe_gene_set)]
unmatched = all_interactors[~all_interactors.isin(pombe_gene_set)]
print(f"  Matched to pombe gene list:     {len(matched):,}")
print(f"  Not matched (e.g. '-' or other organism IDs): {len(unmatched):,}")

# For matched genes, retrieve gene_category
matched_df = (
    pombe_genes[pombe_genes["gene_systematic_id"].isin(matched)]
    [["gene_systematic_id", "gene_category"]]
    .drop_duplicates()
)

# Summary table: count of unique BioGRID interactors per gene_category
category_summary = (
    matched_df.groupby("gene_category", observed=True)
    .size()
    .reset_index(name="unique_interactor_genes")
    .sort_values("unique_interactor_genes", ascending=False)
)
print("\nUnique BioGRID interactor genes by gene_category:")
category_summary
```

```python
throughput_counts = biogrid_df.loc[:,'Throughput'].value_counts()
print("The percentage of reactions that are high throughput is: {:.1f}%".format(throughput_counts['High Throughput']/len(biogrid_df) * 100))

print("The publication sources that contributed to high throughput interactions are:")
high_throughput_sources = biogrid_df.loc[biogrid_df['Throughput'] == 'High Throughput', 'Publication.Source'].value_counts()
print(len(high_throughput_sources), " unique publication sources")
print(high_throughput_sources)

print(sum(high_throughput_sources), " total high throughput interactions")
```

```python
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches

# ── Colour settings ────────────────────────────────────────────────────────────
FG_COLOR = "#2E86AB"   # foreground slice (genes WITH a BioGRID interaction)
BG_COLOR = "#D9D9D9"   # background slice (genes WITHOUT a BioGRID interaction)
FIGURE_BG = "#FFFFFF"  # figure background
# ──────────────────────────────────────────────────────────────────────────────

# ── Numbers for subplot 1: all S. pombe genes ─────────────────────────────────
total_pombe = len(pombe_genes)
n_with_interaction_all = len(matched)          # unique pombe genes in BioGRID
n_without_interaction_all = total_pombe - n_with_interaction_all

# ── Numbers for subplot 2: priority unstudied genes only ─────────────────────
priority_genes = pombe_genes[pombe_genes["gene_category"] == "Priority Unstudied"]
total_priority = len(priority_genes)
n_with_interaction_pug = priority_genes["gene_systematic_id"].isin(matched).sum()
n_without_interaction_pug = total_priority - n_with_interaction_pug

# ── Plot ──────────────────────────────────────────────────────────────────────
fig, axes = plt.subplots(1, 2, figsize=(11, 5.5), facecolor=FIGURE_BG)
fig.suptitle("BioGRID PPI coverage in S. pombe", fontsize=14, fontweight="bold", y=1.01)

pie_configs = [
    {
        "ax": axes[0],
        "values": [n_with_interaction_all, n_without_interaction_all],
        "total": total_pombe,
        "title": "All S. pombe genes",
    },
    {
        "ax": axes[1],
        "values": [n_with_interaction_pug, n_without_interaction_pug],
        "total": total_priority,
        "title": "Priority Unstudied Genes",
    },
]

for cfg in pie_configs:
    ax = cfg["ax"]
    vals = cfg["values"]
    pct = vals[0] / cfg["total"] * 100

    wedges, texts, autotexts = ax.pie(
        vals,
        colors=[FG_COLOR, BG_COLOR],
        autopct="%1.1f%%",
        startangle=90,
        wedgeprops={"edgecolor": FIGURE_BG, "linewidth": 1.5},
        textprops={"fontsize": 11},
    )
    autotexts[0].set_color("white")
    autotexts[0].set_fontweight("bold")

    ax.set_title(cfg["title"], fontsize=12, fontweight="bold", pad=12)
    ax.set_facecolor(FIGURE_BG)

    legend_handles = [
        mpatches.Patch(color=FG_COLOR, label=f"With BioGRID PPI  (n={vals[0]:,})"),
        mpatches.Patch(color=BG_COLOR, label=f"No BioGRID PPI     (n={vals[1]:,})"),
    ]
    ax.legend(handles=legend_handles, loc="lower center",
              bbox_to_anchor=(0.5, -0.12), fontsize=9, frameon=False)

plt.tight_layout()
plt.savefig(
    BASE_DIR / "intermediate_files" / "biogrid_ppi_coverage_pie.png",
    dpi=150, bbox_inches="tight", facecolor=FIGURE_BG
)
plt.show()
print("Figure saved to intermediate_files/biogrid_ppi_coverage_pie.png")
```

## S. cerevisiae BioGRID interactions and pombe orthologs

Repeat the coverage analysis using the S. cerevisiae BioGRID physical interactions dataset.

**Workflow:**
1. Load the S. cerevisiae BioGRID xlsx (data is stored as tab-delimited text in a single Excel column — parsed accordingly).
2. Build a reverse map: cerevisiae gene → S. pombe ortholog(s), using the same parsing logic as Step 5 for human orthologs.
3. Quantify how many S. pombe genes with a cerevisiae ortholog have ≥1 interaction in the cerevisiae BioGRID dataset, broken down by gene category (Studied / Priority Unstudied / Non-Priority Unstudied).

```python
import io
import re
from collections import defaultdict

# ── Paths ─────────────────────────────────────────────────────────────────────
CEREV_BIOGRID_PATH  = BASE_DIR / "ppis" / "biogrid" / "BIOGRID_cerevisiae_physical_interactions.xlsx"
CEREV_ORTHOLOG_PATH = BASE_DIR / "ppis" / "biogrid" / "pombe_cerevisiae_orthologs_20260609.txt"

# ── Load cerevisiae BioGRID ───────────────────────────────────────────────────
# The xlsx stores all data as tab-delimited text in column A; parse via StringIO
raw_xl = pd.read_excel(CEREV_BIOGRID_PATH, header=None, dtype=str)
tsv_text = "\n".join(raw_xl.iloc[:, 0].dropna().astype(str))
cerev_biogrid_df = pd.read_csv(io.StringIO(tsv_text), sep="\t")

print(f"Cerevisiae BioGRID interactions loaded: {len(cerev_biogrid_df):,} rows")
print(f"Columns: {list(cerev_biogrid_df.columns[:8])} ...")
```

```python
# ── Build reverse map: cerevisiae gene → S. pombe ortholog(s) ────────────────
# ugt.parse_ortholog_gene_field handles |, +, and (C)/(N) notation
cerev_to_pombe = defaultdict(list)

with open(CEREV_ORTHOLOG_PATH, encoding="utf-8") as fh:
    for line in fh:
        line = line.rstrip("\n")
        if not line.strip():
            continue
        parts = line.split("\t")
        if len(parts) < 2:
            continue
        pombe_gene  = parts[0].strip()   # kept exactly as-is (incl. fusion notation)
        cerev_field = parts[1].strip()

        for cerev_name in ugt.parse_ortholog_gene_field(cerev_field):
            if pombe_gene not in cerev_to_pombe[cerev_name]:
                cerev_to_pombe[cerev_name].append(pombe_gene)

print(f"Unique cerevisiae gene names in reverse map : {len(cerev_to_pombe)}")
print(f"Total (cerevisiae, pombe) pairs             : {sum(len(v) for v in cerev_to_pombe.values())}")

# Preview entries with >1 pombe ortholog
multi_pombe_cerev = {c: ps for c, ps in cerev_to_pombe.items() if len(ps) > 1}
print(f"\nCerevisiae genes with >1 S. pombe ortholog  : {len(multi_pombe_cerev)}")
for cerev, pombes in sorted(multi_pombe_cerev.items())[:10]:
    print(f"  {cerev:15s} → {pombes}")
```

```python
# ── Re-use pombe_genes already loaded in cell 6 ──────────────────────────────
# (pombe_genes has columns: gene_systematic_id, gene_category, ...)
pombe_id_to_category = dict(zip(pombe_genes["gene_systematic_id"],
                                 pombe_genes["gene_category"]))
print(f"Pombe gene list: {len(pombe_id_to_category):,} genes")
print("Categories:", pombe_genes["gene_category"].value_counts().to_dict())

# ── Unique cerevisiae interactors found in BioGRID ───────────────────────────
# Use Systematic.Name columns (ORF names like YBR265W) to match the ortholog file
cerev_interactors_in_biogrid = set(
    pd.concat([
        cerev_biogrid_df["Systematic.Name.Interactor.A"],
        cerev_biogrid_df["Systematic.Name.Interactor.B"],
    ]).dropna()
)
# Remove non-gene placeholders (e.g. "-")
cerev_interactors_in_biogrid = {g for g in cerev_interactors_in_biogrid if g != "-"}
print(f"\nUnique cerevisiae interactor IDs in BioGRID: {len(cerev_interactors_in_biogrid):,}")

# ── For each S. pombe gene that has a cerevisiae ortholog, check BioGRID coverage
# Build pombe gene → set of cerevisiae orthologs
pombe_to_cerev = defaultdict(set)
for cerev_gene, pombe_genes_list in cerev_to_pombe.items():
    for pg in pombe_genes_list:
        pombe_to_cerev[pg].add(cerev_gene)

# All pombe genes that have at least one cerevisiae ortholog
pombe_with_cerev_ortholog = set(pombe_to_cerev.keys())
print(f"S. pombe genes with ≥1 cerevisiae ortholog : {len(pombe_with_cerev_ortholog):,}")

# Which of those have their cerevisiae ortholog present in BioGRID?
pombe_cerev_in_biogrid = {
    pg for pg, cerevs in pombe_to_cerev.items()
    if cerevs & cerev_interactors_in_biogrid
}
print(f"  …with ≥1 cerevisiae ortholog in BioGRID  : {len(pombe_cerev_in_biogrid):,}")

# ── Coverage summary broken down by gene_category ────────────────────────────
rows = []
for category in ["Studied", "Priority Unstudied", "Non-Priority Unstudied"]:
    cat_genes_with_orth = {
        pg for pg in pombe_with_cerev_ortholog
        if pombe_id_to_category.get(pg) == category
    }
    cat_genes_in_biogrid = cat_genes_with_orth & pombe_cerev_in_biogrid
    total_cat = (pombe_genes["gene_category"] == category).sum()
    rows.append({
        "gene_category": category,
        "total_pombe_genes": total_cat,
        "pombe_genes_with_cerev_ortholog": len(cat_genes_with_orth),
        "pombe_genes_cerev_ortholog_in_BioGRID": len(cat_genes_in_biogrid),
        "pct_of_ortholog_genes_in_BioGRID": (
            round(len(cat_genes_in_biogrid) / len(cat_genes_with_orth) * 100, 1)
            if cat_genes_with_orth else 0.0
        ),
    })

coverage_df = pd.DataFrame(rows)
print("\nS. cerevisiae BioGRID PPI coverage for S. pombe genes (by category):")
coverage_df
```
